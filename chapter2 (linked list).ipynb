# Doubly Linked List

```python
# DList 특
# Head<->item <-> item <-> tail
# usually they look like this^

class DList:
    class Node:
        # Making Node w prev and link
        def __init__(self, item, prev, link):
            self.item = item
            self.prev = None
            self.link = None

        def __init__(self, item, prev, link):
            # Nothing Nothing Nothing
            self.head = self.Node(None,None,None)
            # Connect tail to head 
            self.tail = self.Node(None, self.head, None)
            # Connect head to tail
            self.head.next = self.tail
            # there is nothing on the list 
            self.size = 0

        def size(self): return self.size
        def is_empty(self): return self.size == 0

        def insert_before(self, p, item):
            t = p.prev
            # prev, item, p.next
            n = self.Node(item, t, p)

            #  let's say DList is A B C
            # A -> B
            p.prev = n

            # C -> B
            t.next = n
            # increase size
            self.size += 1

        def insert_after(self, p, item):
            t = p.next
            # prev is p then t comes after
            n = self.Node(item, p, t)

            # p -> item
            p.next = n
            # t -> item
            t.prev = n

            self.size += 1

        def delete(self, x):
            # x's previous and next
            f = x.prev
            r = x.next

            # ignore x but connect f and r
            f.next = r
            r.prev = f

            # x is gone
            self.size -= 1
            return x.item


        def print_list(self):
            if self.is_empty():
                print("Empty llist")
            else:
                # p = item cuz DList has head and tail that is empty
                # Head <-> item <-> tail
                p = self.head.next

                # if p reaches tail then printing list should end cuz tail is the end of DList
                while p != self.tail:
                    # same here
                    if p.next != self.tail:
                        print (p.item, " <=>", end="")
                    # if it is tail then we don't want to add extra <=> yeye
                    else: 
                        print (p.item)
                    p = p.next


        
```

```python
def f(n):
    if n ==0:
        return 
    print(n%2, end="")
    f(n//2)

f(12)
```

```python
# 1.50

def factorial(n, fact):
    if n == 1:
        return fact
    return factorial(n - 1, fact * n)
```

```python
def factorial (n):
    if n == 1:
        return n
    else:
        # here is the key point
        return n * factorial(n-1)

factorial(4)
```

```python
def fibo(n):
    if n <= 1:
        return n
    return fibo(n - 1) + fibo(n - 2)

fibo(11)
```

```python
def to20 (n):
    if n == 20:
        return print(20)
    print (n, end=" ")
    return to20(n+1)

to20(1)
```

```python
def count_digit (n):
    count = 0
    if n == 0:
        return 0
    
    count += 1

    return count + count_digit(n // 10)

count_digit(1234123)
```

```python
def jwpower (x, m): #x^m
    if m == 0:
        return 0
    elif m == 1:
        return x

    return x * jwpower (x, m-1)

jwpower(2,10)
```

```python
def hanoi(n, start, end, temp):
    if n == 1:
        # 시작에서 끝까지 이동
        print (start, "->", end)
        return
    else: 
        hanoi(n-1, start, temp, end) 
        print (start, "->", end)
        hanoi (n-1, temp, end, start)

# thinking about this problem with 2 disk make it easier
# move start to temp
# then temp to end
# if n == 1 -> just go to end

# goat yt vid https://www.youtube.com/watch?v=ShRHZKpYe4w

hanoi (3, "A", "C", "B")  
```

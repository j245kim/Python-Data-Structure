```python
def f(n):
    if n ==0:
        return 
    print(n%2, end="")
    f(n//2)

f(12)
```

```python
# 1.50

def factorial(n, fact):
    if n == 1:
        return fact
    return factorial(n - 1, fact * n)
```

```python
def factorial (n):
    if n == 1:
        return n
    else:
        # here is the key point
        return n * factorial(n-1)

factorial(4)
```

```python
def fibo(n):
    if n <= 1:
        return n
    return fibo(n - 1) + fibo(n - 2)

fibo(11)
```

```python
def to20 (n):
    if n == 20:
        return print(20)
    print (n, end=" ")
    return to20(n+1)

to20(1)
```
